```python
import pandas as pd
```

```python
# 1. Load the file
df1 = pd.read_csv("final_bengali_dataset.csv")  
df2 = pd.read_csv("Dhoroni_Title_Content.csv")
```

```python
# 2. Combine them
# Use concat to stack them vertically (on top of each other)
final_df = pd.concat([df1, df2], ignore_index=True)

# 3. Save the result back to a new CSV file
final_df.to_csv("merged_bengali_dataset.csv", index=False, encoding="utf-8")

print("Files successfully merged into merged_dataset.csv!")
print(f"Total rows in final file: {len(final_df)}")
```
