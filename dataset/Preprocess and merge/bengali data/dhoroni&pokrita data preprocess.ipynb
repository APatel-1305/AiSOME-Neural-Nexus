```python
CLIMATE_KEYWORDS = [k.lower() for k in CLIMATE_KEYWORDS]
```

```python
science["text"] = science["text"].fillna("").astype(str)

national["text"] = national["text"].fillna("").astype(str)
```

```python
import re

def clean(text):

    text = str(text).lower()

    text = re.sub(r"\s+", " ", text)

    return text.strip()

science["text"] = science["text"].apply(clean)
national["text"] = national["text"].apply(clean)
dhoroni["text"] = dhoroni["text"].apply(clean)
```

```python
def climate_article(text):

    for kw in CLIMATE_KEYWORDS:

        if kw in text:

            return True

    return False
```

```python
science_filtered = science[
    science["text"].apply(climate_article)
]

national_filtered = national[
    national["text"].apply(climate_article)
]
```

```python
print(len(science))
print(len(science_filtered))

print(len(national))
print(len(national_filtered))
```

```python
final_df = pd.concat(

    [
        dhoroni,
        science_filtered,
        national_filtered
    ],

    ignore_index=True
)
```

```python
final_df = final_df.drop_duplicates(
    subset=["text"]
)
```

```python
final_df = final_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
final_df.to_csv(
    "Climate_Pretraining_Corpus.csv",
    index=False,
    encoding="utf-8-sig"
)
```

### Shortening the `Climate_Pretraining_Corpus.csv` to 2300 random rows

```python
import pandas as pd

# Load the existing CSV
df_full = pd.read_csv('Climate_Pretraining_Corpus.csv', encoding='utf-8-sig')

# Sample 2300 random rows
df_short = df_full.sample(n=2300, random_state=42).reset_index(drop=True)

# Save the shortened DataFrame to a new CSV
df_short.to_csv('Climate_Pretraining_Corpus_2300_rows.csv', index=False, encoding='utf-8-sig')

print(f"Original corpus size: {len(df_full)} rows")
print(f"Shortened corpus size: {len(df_short)} rows")
print("Saved shortened corpus to 'Climate_Pretraining_Corpus_2300_rows.csv'")

display(df_short.head())
```
