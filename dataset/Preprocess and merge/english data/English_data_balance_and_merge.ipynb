```python
import pandas as pd
```

```python
df = pd.read_csv("final merge_dataset.csv")
df1 = pd.read_csv("twitter_dataset.csv")

print(df.head())
print(df1.head())
```

```python
df.info()
```

```python
df1.info()
```

```python
# In df1 replace Nan By NONE in sentiment column
df1['sentiment'].fillna('NONE', inplace=True)
print(df1.head())
```

```python
# in df1 upper Favor to FAVOR and against to AGAINST in sentiment
df1['sentiment'] = df1['sentiment'].replace({'Favor': 'FAVOR', 'Against': 'AGAINST'})
print(df1.head())
```

```python
df1.info()
```

```python
# remane columns in df1 clean_text to sentence and sentiment to label
df1 = df1.rename(columns={'clean_text': 'sentence', 'sentiment': 'label'})
print(df1.head())
```

```python
#I want to join both the sentence vertically based on column name and drop excess column
df2 = pd.concat([df, df1], axis=0)
print(df.head())
```

```python
#Drop message column
df2 = df2.drop('message', axis=1)
print(df.head())
```

```python
df2.info()
```

```python
# check the label occurances
df2['label'].value_counts()
```

```python
# Drop label NONE and FAVOR count to 4257 each ramdomly

# 1. Separate the categories into their own DataFrames
df_none = df2[df2['label'] == 'NONE']
df_favor = df2[df2['label'] == 'FAVOR']

# Keep all other labels (like 'AGAINST', etc.) completely untouched
df_others = df2[~df2['label'].isin(['NONE', 'FAVOR'])]

# 2. Randomly sample exactly 4,257 rows from NONE and FAVOR
# (random_state ensures your code produces the exact same random split every time)
df_none_sampled = df_none.sample(n=4257, random_state=42)
df_favor_sampled = df_favor.sample(n=4257, random_state=42)

# 3. Vertically concatenate everything back together
balanced_df = pd.concat([df_none_sampled, df_favor_sampled, df_others])

# 4. Shuffle the dataset so the rows aren't grouped by label
balanced_df = balanced_df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
# download balanced df as csv
balanced_df.to_csv('English_train_data.csv', index=False)
```
