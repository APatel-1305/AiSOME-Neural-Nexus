```python
import pandas as pd
```

```python
df = pd.read_csv("train.csv")
```

```python
df.head()
```

```python
# Combine headline and content
df["text"] = (
    df["headline"].fillna("") + " " +
    df["article"].fillna("")
)

# Remove extra spaces
df["text"] = (
    df["text"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Remove duplicate articles
df = df.drop_duplicates(subset=["text"])
```

```python
keywords = [
    "जलवायु",
    "मौसम",
    "बारिश",
    "तापमान",
    "गर्मी",
    "ठंड",
    "सूखा",
    "बाढ़",
    "चक्रवात",
    "तूफान",
    "हिमपात",
    "मानसून",
    "ग्लोबल वार्मिंग",
    "पर्यावरण",
    "प्रदूषण",
    "ग्रीनहाउस",
    "कार्बन",
    "समुद्र",
    "समुद्र स्तर",
    "आईएमडी",
    "मौसम विभाग"
]
```

```python
pattern = "|".join(keywords)

df = df[
    df["text"].str.contains(pattern, case=False, na=False)
]
```

```python
df[["text"]].to_csv(
    "hindi_mlm_dataset2.csv",
    index=False,
    encoding="utf-8-sig"
)

print(df.shape)
```

```python
dataset1 = pd.read_csv("hindi_mlm_dataset2.csv")
dataset2 = pd.read_csv("hindi_mlm_dataset3.csv")
```

```python
target_total = 4600 #rows

len1 = len("hindi_mlm_dataset3.csv")
len2 = len("hindi_mlm_dataset2.csv")
total_len = len1 + len2
print(total_len)

n1 = 830
n2 = target_total-n1

sample1 = dataset2.sample(n=n1, random_state=42)
sample2 = dataset1.sample(n=n2, random_state=42)

final_dataset = pd.concat([sample1, sample2], ignore_index=True)

print(final_dataset.head())
print(final_dataset.info())
```

```python
final_dataset.to_csv(
    "final_dataset_hindi1.csv",
    index=False
)

print(final_dataset.shape)
```
